Synthetic survey dataset; data is simulated for portfolio purposes.

```python
import pandas as pd
df = pd.read_csv('../data/cleaned_survey_data.csv')
df.head()
```

```python
df.describe(include='all').T
```

```python
df.groupby('preferred_platform').agg(respondents=('response_id','count'), avg_satisfaction=('overall_satisfaction','mean'), avg_repeat_intent=('repeat_purchase_intent','mean')).round(2)
```

```python
drivers=['delivery_satisfaction','product_quality','reviews_importance','return_experience','customer_support','overall_satisfaction']
df[drivers].corr()['overall_satisfaction'].sort_values(ascending=False)
```
